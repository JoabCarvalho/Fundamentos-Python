#### Exercício 7

Receba um número inteiro do usuário e use o operador de módulo (%) para verificar se o número é par ou ímpar. Imprima o resultado.

Obs: Não estudamos ainda os condicionais, mas pesquise se necessário e tente resolver.

```python
# Solução

# Criamos um laço de repetição que vai rodar para sempre...
while True:
    # O programa vai "tentar" fazer o que está aqui dentro
    try:
        # Pede a entrada do usuário e tenta transformar em número inteiro (int)
        nInt = int(input("Digite um número inteiro: "))
        
        # Se a transformação deu certo, nós saímos do laço com o 'break'
        break
        
    # Se o usuário digitou letras, o Python gera um erro chamado ValueError.
    # Nós capturamos esse erro aqui para o programa não fechar!
    except ValueError:
        # Avisamos o usuário e o laço recomeça, pedindo o número de novo
        print("Ops! Isso não é um número inteiro válido. Tente novamente.")

# Quando sai do laço, significa que temos um número correto guardado em nInt
print(f"Muito bem! Você digitou o número inteiro: {nInt}")

if nInt % 2 == 0:
    print(f"O número {nInt} é par.")
else:
    print(f"O número {nInt} é ímpar.")
```

#### Exercício 30

Escreva um programa que peça ao usuário para digitar sua altura em metros (ex: 1.75) e seu peso em quilogramas (ex: 68.5). Calcule o Índice de Massa Corporal (IMC) usando a fórmula IMC = peso / (altura * altura) e imprima o resultado formatado com duas casas decimais.

```python
# Solução

while True:
    try:
        # 1. Pegamos a entrada como texto (string) primeiro
        entrada = input("Digite sua altura em metros: ")
        
        # 2. Se o usuário usou vírgula (ex: 1,70), trocamos por ponto (1.70)
        entrada_tratada = entrada.replace(",", ".")
        
        # 3. Se o usuário digitou um número inteiro sem ponto (ex: 170 ou 180)
        if "." not in entrada_tratada:
            # Convertemos para float e dividimos por 100 (ex: 170 vira 1.7)
            altura = float(entrada_tratada) / 100
        else:
            # Se já tem ponto, apenas convertemos diretamente
            altura = float(entrada_tratada)
            
        # Se tudo deu certo até aqui, saímos do loop!
        break
        
    except ValueError:
        # Se o usuário digitou letras ou algo que o Python não consegue converter de jeito nenhum
        print("Ops! Isso não parece uma altura válida. Tente novamente usando apenas números.")

while True:
    try:
        # Recebe o peso do usuário como string
        peso = input("Digite seu peso em Kg: ")
        # Altera vírgula para ponto, caso o usuário tenha digitado assim
        peso = peso.replace(",", ".")
        # Tenta converter para float
        peso = float(peso)
        break
    except ValueError:
        # Se o usuário digitou letras ou algo que o Python não consegue converter de jeito nenhum
        print("Ops! Isso não é um peso válido. Tente novamente.")

# Calculando o IMC
imc  = peso / (altura ** 2)

# Classificando o IMC
if imc < 18.5:
    classificacao = "Abaixo do peso"
elif 18.5 <= imc < 25:
    classificacao = "Peso normal"
elif 25 <= imc < 30:
    classificacao = "Sobrepeso"
else:
    classificacao = "Obesidade"

# Exibindo o resultado
print(f"Baseado em sua altura de {altura:.2f} m e peso de {peso:.2f} Kg")
print(f"Seu IMC é: {imc:.2f}. Classificação: {classificacao}.")
```
